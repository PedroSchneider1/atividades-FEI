```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("notas.csv")
print(df.head(5))

x = []
for i in df.values:
  x.append([i[0] ,i[1]])

x1 = df["livros"]
x2 = df["aulas"]
y = df["notas"]

# Plot feito com chatgpt :)
fig = plt.figure(figsize=(11, 8))
ax = fig.add_subplot(111, projection="3d")

# Scatter plot
scatter = ax.scatter(
    x1,
    x2,
    y,
    c=y,                 # Color based on grade
    cmap="viridis",
    s=100,
    alpha=0.85,
    edgecolors="black",
    linewidth=0.6
)

# Colorbar
cbar = fig.colorbar(scatter, ax=ax, pad=0.1, shrink=0.7)
cbar.set_label("Nota", fontsize=11)

# Axis labels
ax.set_xlabel("Livros lidos", fontsize=11, labelpad=10)
ax.set_ylabel("Frequência às aulas", fontsize=11, labelpad=10)
ax.set_zlabel("Nota", fontsize=11, labelpad=10)

# Better viewing angle
ax.view_init(elev=25, azim=-125)

# Grid
ax.grid(True, alpha=0.3)

# Improve tick appearance
ax.tick_params(axis="both", labelsize=9)

plt.tight_layout()
plt.show()
```

```python
#Divide o dataset em treino e teste
X_train, X_test, Y_train, Y_test = train_test_split(x,y, test_size=0.2, random_state=42)

#Faz a regressao
modelo = LinearRegression()
modelo.fit(X_train,Y_train)

# (livros, aulas)
print(modelo.predict([[2,11]]))

print(modelo.predict([[0,5]]))

print(modelo.predict([[4,20]]))

print(modelo.predict([[2,10]]))

print(modelo.predict([[4,15]]))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

df = pd.read_csv("agua_alpes.csv")

X = df["BPt"]          # 1D inicialmente
y = df["Pressure"]     # 1D

# Transforma X em 2D
X = np.array(X).reshape(-1, 1)
y = np.array(y)

plt.scatter(X, y)
plt.xlabel("BPt")
plt.ylabel("Pressão")

modelo = LinearRegression()
modelo.fit(X, y)

# Pressão para 200°
print(modelo.predict([[200]]))

# Pressão para 215°
print(modelo.predict([[215]]))

plt.show()
```

```python
a = modelo.coef_[0]
b = modelo.intercept_

print("Coef Angular:", a)
print("Coef Linear:", b)

# Previsões para todos os X
previsao = modelo.predict(X)

# Gráfico
plt.scatter(X, y, color="red", label="Dados")

# Ordenar X para a reta não ficar "ziguezagueando"
plt.plot(X, previsao, color="blue", label="Regressão Linear")

plt.xlabel("BPt")
plt.ylabel("Pressure")
plt.legend()
plt.show()
```
